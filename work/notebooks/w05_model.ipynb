# ML-08 — Train and Compare the Model

This notebook trains and compares three machine learning classifiers using a client-holdout split to predict content decline.

## 1. Split data (client-holdout)

```python
import pandas as pd
import numpy as np
from pathlib import Path
import sys
sys.path.append(str(Path('../../scripts')))
import subprocess

results_path = Path('../../../outputs/model_results.json')
if not results_path.exists():
    subprocess.run([sys.executable, '../../scripts/03_train_model.py'])

import json
with open(results_path) as f:
    results = json.load(f)
print(f"Split strategy: {results['split_strategy']}")
print(f"Train size: {results['train_rows']:,} | Test size: {results['test_rows']:,}")
```

## 2. Train classifiers & Compare

```python
print(f"Best model found: {results['best_model']['name']}")
print("\nValidation metrics comparison:")
models_df = pd.DataFrame(results['models']).T
baseline_s = pd.Series(results['baseline'])
print(models_df[['roc_auc', 'average_precision', 'precision_at_50', 'recall', 'f1']])
```

## 3. Interpret Feature Importance

```python
print("Top 10 feature importances for best model:")
feat_imp = pd.DataFrame(results['best_model']['feature_importance_top'])
print(feat_imp.head(10))
```
